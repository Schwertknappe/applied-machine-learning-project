### Discussion

#### **1. Which models performed best overall, and on which feature sets?**
The top-performing model was the `Random Forest` classifier trained on the **raw** time-series dataset with an f1 score of ~0.84, a recall score of ~0.84 and a ROC AUC score of ~0.93 (see Table 1).


#### **2. Were there significant differences in performance between using engineered features and raw time-series data? Why might this be?**
In general models trained on raw time-series data outperformed thos trained on engineered features, with consistent differences of ~0.01–0.02 across all metrics. This performance gap, supported by the low standard deviations and visible in Figure 1, is likely non-random and attributable to the loss of high-resolution temporal information during feature extraction. Such information may contain subtle yet important distinctinguishing patterns between weld quality classes. However, the fact that models trained on engineered features still performed relatively well suggests that some extracted variables retained meaningful predictive power. 

Notably, as the only exception, the SVM classifier outperformed its raw-data counterpart when trained on engineered features. This may be attributed to its tendency to overfit with high-dimensional data, making it more suited for reduced, structured input.


#### **3. How do the chosen evaluation metrics reflect the models' ability to distinguish between good and bad quality welds?**

The chosen evaluation metrics all reflect a models' ability to correctly identify a good and bad weld. In all metrics the `Random Forest` classifier trained on the raw time-series dataset displays scores higher than the other two best performing models. This is reflected in its performance of correctly identifying qualtiy class in ~84.39% of welds. The next best performing model in all metrics - the `Gradient Boosting` model trained on the raw time-series dataset - correctly identifies quality class in ~82.99% of welds, whereas the third best performing model - `Random Forest` trained on engineered features - is able to correctly identify quality class in ~82.34% of welds.

While the evaluation metrics provide a general view of performance, they do not differentiate between the types of classification errors. For example as seen in Figure 2, the `Gradient Boosting` model, despite strong overall scores, exhibited a higher number of false positives (3454) than false negatives (1922), which could be critical in a production context. Since false positives imply faulty welds being mistakenly accepted and potentially causing more costly issues further down the production chain, the chosen metrics alone may not fully reflect real-world risk.

## Exercise 6.4: Conclusion and Future Work

Several models displayed robust prediction performance, especially when trained on raw time-series data. The best performing models were the `Random Forest` classifier trained both on raw and engineered features, as well as the `Gradient Boosting` classifier trained on raw time-series data.

The stronger performance of models trained on raw time-series data over engineered features (as seen in Figure 1) suggests that there may be time related data patterns instrumental in predicting weld quality. Nevertheless some engineered features also display promising performance, being only slightly behind raw data in prediction quality metrics.

The engineered features used may have been overly generalized, potentially omitting signal characteristics critical for distinguishing weld quality. Future work should explore a broader and more domain-informed set of features, such as peak arc frequency, amplitude modulation, or weld duration variability. Furthermore, experts should be consulted for vital variables that may already be known to show distinct patterns in faulty welds.

This exercise was limited to using classifiers found in the `scikit-learn` library, however this library does not contain native support for directly modeling time-series data without first flattening data. Future work consider frameworks designed for temporal data, such as `sktime`, `TensorFlow` or `PyTorch`. This may lead to an even more improved prediction performance of models trained on raw data.  Additionally, there is a selection of gradient boosting algorithms outside of `scikit-learn` that may show better ability to handle large datasets, e.g. `lightgbm`.

Lastly, it is possible that different models may be better at identifying different error types in faulty welds. Thus, future work should explore constructing ensembles of multiple different models to reduce classification errors via majority votes.
